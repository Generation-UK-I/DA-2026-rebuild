APIs can return JSON data, i.e. key:value pairs containing text and numerical data.

We can ask our CaaS app to return information about the cat image as JSON rather than the image, because the cataas endpoint supports returning metadata as JSON by adding the `?json=true` query parameter.

Run the code below and ensure that you receive output similar to: `{ "_id": "595f280e557291a9750ebf4d", "tags": ["cute"], "createdAt": "2017-01-08T01:10:22.001Z" }`

````python
import requests
url = "https://cataas.com/cat?json=true"
response = requests.get(url)
print(f"Status code: {response.status_code}")
data = response.json()
print(data)
````

## Python Requests and JSON Libraries

Python's `requests` and `json` libraries are essential when working with APIs, web services, and data pipelines.

- **requests** lets Python communicate with web servers using HTTP. You can retrieve data, submit forms, send JSON, and interact with REST APIs.
- **json** lets Python convert JSON data into Python objects and convert Python objects into JSON text.

Together, they allow you to write Python code which retrieves data from external services, process it, and send results elsewhere.

### JSON

JSON (JavaScript Object Notation) is a text format for exchanging structured data. Despite its name, it is widely used with Python, Java, JavaScript, and many other languages.

Example:

```json
{
    "name": "Alice",
    "age": 28,
    "active": true,
    "skills": ["Python", "SQL", "Linux"],
    "address": null
}
```

The structure and syntax of a JSON object is identical to a Python dictionary; A dictionary is a data type which exists in memory, whereas JSON is a data format comprised of strings, that exists in files that are in storage or in transit.

We can convert between them using methods from the `json` library, which is part of the Python Standard Library, so does not need to be installed, just imported:

- Use `json.dumps(dict)` to transform dict --> json
- Use `json.loads(json)` to transform json --> dict

### Requests

The `requests` library is an external package so requires installation before use: `pip install requests`

You can then import it into your python files like any other library.

The `requests` library provides a straightforward interface for making HTTP requests. Its methods correspond to the common HTTP methods used by web APIs.

|Method|Usage|
|---|---|
|`requests.get()`|Retrieve data	e.g. Get products or customer records|
|`requests.post()`|Submit data e.g. Create a new order|
|`requests.put()`|Replace a resource e.g. Replace a complete customer record|
|`requests.patch()`|Partially update a resource e.g. Change a customer's email address|
|`requests.delete()`|Delete a resource e.g. Delete an order|

We've seen code similar to this already in our CaaS app, but it's worth revisiting now we have a little more context:

````python
import requests

# 1. Define the API endpoint
url = "https://jsonplaceholder.typicode.com/users/1"
 
# 2. Make a GET request to the API
response = requests.get(url)

# 3. Check the request was successful
print(f"Status code: {response.status_code}")
print(response.text)

# 4. parse the response as a json object to the 'data' variable
data = response.json()
````

>JSONPlaceholder is a free online REST API that you can use whenever you need some fake data: in documentation and tutorials, in code examples, in a quick prototype, or simply to test things locally.

The code above demonstrates the creation of a **requests.Response** object, which is the typical way we interact with the HTTP response received from the server. We can call a number of different methods against this object, some examples include:

- `status_code`: returns the HTTP status code (reviewed previously, 200=OK)
- `url`: Returns the URL of the response
- `headers`: Returns a dictionary of response headers
- `text`: Returns the content of the response, in unicode (used above)
- `json()`: Returns a JSON object of the result (if the result was written in JSON format, used above)

More response.Object methods can be reviewed here: [w3schools](https://www.w3schools.com/python/ref_requests_response.asp)
