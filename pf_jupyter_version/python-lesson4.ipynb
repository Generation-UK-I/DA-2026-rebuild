# Functions

The next feature we need to look at is Functions, which is one of the most common approaches to writing and structuring your code.

Functions allow you to write individual code blocks which perform a specific action, and separate those blocks so that they can be called when required. This is in contrast to writing lines one after another, and then running one line at a time sequentially from top to bottom.

A function is created with the def keyword, followed by a descriptive function name, for example:

```py
def hello_world():
```

When you run the function it executes code that is indented within the function:

```py
def hello_world():
    print('hello world')
```

>**Notice**: The lines of code to be executed are indentated from the function definition by 1x `TAB`.

In order to exectute the function we need to call it, which we do by stating the function name:

```py
def hello_world():
    print('hello world')

hello_world()
```

You might be asking "why bother?" that is more lines than just `print('hello world')`, but the function can contain many lines of code, and it can be called over and over again:

````python
def hello_world():
    print('hello world')
    print('this is a print statement')
    print('inside a function')

hello_world()
hello_world()
hello_world()
hello_world()
hello_world()
````

## Passing Data into a Function

As mentioned, functions are very common and important in Python.

We've already seen them:

- Print data
- Capture input data
- Append to an existing list

However, we can do a LOT more.

One of the first ways to make our functions more useful is by passing information into it. For example, take the first function we saw:

```py
def hello_world():
    print('hello world')

hello_world()
```

This function works, but it can only do one thing - what if we want to say "hello" to more than just "world"?

To do so we need to change our code, and make it accept external values.

The changes are actually familiar:

```py
def hello_world(name): # Create a variable inside our function, called a 'Parameter'
    print(f'hello {name}') # Use the Parameter by creating an f-string

hello_world('Steve') # Pass a value to the Parameter when calling the function, called an 'Argument'
```

Try it out below, and change the argument to any other strings

````python
def hello_world(name):
    print(f'hello {name}')

hello_world('Steve')
````

>Strictly speaking, the value declared in the function is a **Parameter**, and the value passed through when calling the function is an **Argument**, but you may find them both being referred to as arguments.

This code looks a little complex at first, but it just combines the concepts we've covered so far, and by doing so we open up a lot of functionality.

````python
# An app to create a username for new employees

Users = [] # Start with empty list

def formatted_name(first, last): # Define function and receive 2 arguments
    print(f"Welcome {first.title()} {last.title()}") # Interpolate the arguments
    username = f'{first.title()}.{last.title()}' # Use args to make a new 'username' object
    print(f'Your username is {username}') # Print the new username object
    Users.append(username) # Append the new object to the list
    print(Users) # Print the list to verify

first_name = input('what is your first name?') # Capture first and last names using input()
last_name = input('what is your last name?')

formatted_name(first_name, last_name) # Call the function and pass the name arguments
````

>**Notice**: The "first_name" and "last_name" `input()` statements run before the function is called

## Parameters and Arguments

### Positional Arguments

The previous two examples had you passing multiple arguments into the function. The defaul behaviour is that arguments are passed through in the order that you provide them, so if you have the following function and call:

```py
def my_funct(param1, param2, param3):
    ...

my_funct(arg1, arg2, arg3)
```

Then `arg1` is passed to `param1`, `ar2` to `param2`, and `arg3` to `param3`. We call this behaviour **positional arguments**.

There are a few different ways of passing arguments through to parameters when the default behaviour does not meet your requirements.

### Keyword Arguments

One alternative method is called **Keyword Arguments** (often shortened to kwargs), with this method you can state the name of the parameter your passing an argument for in the format: `function_name(parameter = argument)`

For example, say you want to calculate average customer spend over two weeks.

In the below cell uncomment the function calls one at a time and execute the cell to see the difference between positional and keyword arguments

````python
def avg_spend(week1sales, week2sales, num_custs):
    avg_cust_spend = (week1sales + week2sales)/num_custs
    print(avg_cust_spend)

# avg_spend(425, 5400, 7630)
# avg_spend(num_custs = 425, week1sales = 5400, week2sales = 7630)
````

You can see that the calculations are wildly different if we don't use keyword arguments.

### Arbitrary Arguments

The last example of using arguments we'll review is when we want to pass a flexible number of arguments into our function. For example say we want to calculate the total sales for a day from all of the individual transactions - but we don't know in advance how many transactions there will be:

````python
def daily_sales(*transactions):
  total = 0
  for num in transactions:
    total += num
  return total

day1_total = daily_sales(10, 60, 30, 10, 20, 30, 40)
print(day1_total)

day2_total = daily_sales(40, 50, 30, 20)
print(day2_total)

# Add a day3_totals variable and call daily_sales() with your own values
````

A few things to notice in this code:

- When calling the function the arguments are passed through as a tuple, which is an **iterator**
  - This is the case for the previous examples too, but it's more obvious here due to the for loop
- Our function contains a **for loop** which loops through the arguments, regardless of how many it receives, so the same function works for different daily sales volumes
- The function doesn't print the calculation, instead it **returns** it, and it will be assigned to the variable `day1_total` or `day2_total`, which themselves are printed.

This last point about `return` is particularly important, because we typically have multiple functions doing different things. Often we want the output of one function, to be sent into another for multi-stage processing to occur, as per the next example.

````python
def first_func(value):
    value = value + 10
    print(value) # Not required, just verifying the value is changing
    return value

def second_func(value):
    value = value ** 2
    print(value) # Not required, just verifying the value is changing
    return value

def third_func(value):
    value = value / 3
    print(value) # Not required, just verifying the value is changing
    return value

user_num = int(input('Enter a number'))

first_stage = first_func(user_num)
second_stage = second_func(first_stage)
third_stage = third_func(second_stage)

print(third_stage)
````

## Importing Modules

When we've created a large collection of python functions, for the purpose of DA perhaps we have functions to extract data, functions to clean data, functions to transform data, and so on, it can become difficult to keep track of them, and they can start to make our python (`.py`) files unweildy to work with.

We'll explore this concept by working through an example.

1. First create a new directory called `modules_demo`. In this directory you need the following file [math_functions.py](./modules_demo/math_functions.py) which you should create or download.
2. You can see that this file contains 10x simple functions, although there could be many more, and much more complex; Each of the functions is called at the bottom of the file. Execute the file to confirm all of the functions work.
3. In the `modules_demo` directory create a new file called `calculator.py` and to begin with cut just the first function call from `math_functions.py`: `print(add(5, 3))`, and paste it into `calculator.py`
4. To demo the behaviour that we want you need to comment out all of the remaining function calls in `math_functions.py` (A VSC shortcut is to highlight them all and press `[CTRL] + [?]`)
    - If you do not, then all of the calls in `math_functions.py` will still be executed when you run `calculator.py`
5. If you try to run `calculator.py` you will receive an error message: `NameError: name 'add' is not defined`. This expected because the `add()` function doesn't exist in the new file, but we can import it.
6. At the top of `calculator.py` add the line: `from math_functions import add`, then execute the file again. You should find that the function runs, despite not existing in `calculator.py`.
7. Next return to `math_functions.py` and move the call: `print(power(5, 3))` to `calculator.py` as well (ensure you un-comment it)
8. Update your import statement to read: `from math_functions import add, power` and run the file to see both functions being called from `math_functions.py`
9. Finally move all of the remaining function calls into `calculator.py`, keeping the actual function definitions where they are, and update the import statement to: `from math_functions import *` to import all functions.

You have now seen how you can import a single function, selected functions, or all functions from a separate file.

The file containing the functions we want to use is called a module.

>You can also import the module using simply `import math_functions`, but when you do so you need to state the module name every time you call a function, for example `print(math_function.add(5, 3))`

# Python Libraries

First let's clarify a few terms:

|Term|Definition|
|---|---|
|Module|A single python file, containing functions, classes, or variables|
|Package|A directory containing multiple related modules|
|Library|A collection of related modules and packages providing a range of functionality|

A default installation of Python includes flexible and broad functionality, however many of it's features are modularised and not available immediately. This keeps your python installation lean and fast, because it only loads what you need. To access the extra functionality we can import libraries using an `import library_name` statement, or individual modules from a library using `from library_name import module_name`.

Many different libraries are available within the defaul installation these are referred to as the **Python Standard Library**.

A small selection of some of the most commonly used libraries include:
