# Unit Testing

At its core, unit testing is the practice of writing automated code to verify that individual pieces of your apps (typically functions or classes) behave exactly as expected under various conditions.

There are a range of ways to carry out testing, and as with most things, ideally we want to automate it. One of the most common testing frameworks is called `pytest`.

We should test throughout the development process, in order to catch bugs early, prevent regressions (when new code breaks old features), and document how a piece of code is meant to be used.

The Anatomy of a Test: Most unit tests follow the AAA pattern:

- **Arrange**: Set up the inputs and state.
- **Act**: Call the function or method being tested.
- **Assert**: Check that the output matches the expected result.
  - We have seen many versions of this so far in our exercise notebooks, e.g. `assert result == 42`

Unlike manual testing where you run a Python script or cell in a notebook then visually inspect the output, automated unit tests run in milliseconds, can be executed repeatedly.

## Introduction to PyTest

PyTest is the de-factor standard testing framework in the modern Python ecosystem.

You have already seen and used assert, Pytest builds on top of it, however it's able to run hundreds of assertions at once.

## Pytest with Jupyter Notebooks

Testing within Jupyter notebooks is a little different, because everything is all in one place. We don't create separate `test_*` files or functions. We also do not want to out into the Terminal to run our tests. This is where Ipytest comes in to bridge the gap.

Using ipytest is actually pretty straight forward, adapting the same naming conventions and assert logic from standard pytest, with just a couple of additions. Review the following cell to see the differences:

````python
import ipytest

# Configure ipytest to integrate with the notebook environment
ipytest.autoconfig()

def add_two(x):
    return x + 2

# Write tests using standard pytest naming conventions
def test_add_two():
    assert add_two(3) == 5
    assert add_two(-2) == 0

# Run the tests programmatically in the cell
ipytest.run()
````
