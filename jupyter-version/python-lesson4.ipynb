# Dictionaries

So far we've looked at lists and tuples as containers for our values, dictionaries are another important data structure.

A dictionary is comprised of key-value pairs, separated by a colon `'key':'value'`.

Each KVP is akin to an item in a list, and we can have as many KVPs in our dictionary as we want, separating each one with a comma, just like a list.

````python
# Dictionaries are declared with braces '{ }'
first_dict = {'first_key':'first_value', 'second_key':'second_value'}

# They may be written on one line, but are often written across multiple lines for easier readability
second_dict = {
    'first_key':'first_value',
    'second_key':'second_value'
}
````

The items in the dictionary can be any data type:

````python
cake_dict = {
    'name' : 'Chocolate Fudge Cake',
    'servings' : 12,
    'layers' : 4,
    'frosting' : True,
    'ingredients' : ['flour', 'milk', 'eggs', 'butter', 'cocoa powder', 'sugar'],
    'allergens':{'wheat':True,'milk':True,'eggs':True,'Peanuts':False}
}
````

The above example also demonstrates how dictionaries are regularly used to model real-world objects. Python's flexibility means it can be used in a wide range of scenarios, so the above structure might be used to create a bakery app. However, in data analysis we're much more likely to find an example like the following, containing important data as values, along side keys which provide context.

````python
sales_by_store = {
    "New York" : 72450,
    "Paris" : 65948,
    "Peckham" : 243763
}
````

## Modifying Dictionaries

Like lists, dictionaries are mutable, i.e. we can add, remove, and modify key-value pairs, and they are ordered (but not indexed).

### Populating an Empty Dictionary

The syntax to add a new key-value pairs is as follows: `dictionary_name[new_key] = new_value`

````python
# Start with an empty dictionary
daily_sales = {}

# call the dictionary, provide the new key in square brackets, then the value.
daily_sales['monday'] = 1430
daily_sales['tuesday'] = 1645
...
daily_sales['sunday'] = 'closed'

# Print the dictionary to verify
print(daily_sales)
````

The same syntax is used to update existing key-value pairs also: `dictionary_name[existing_key] = new_value`

````python
# The factory decided to open on Sunday
daily_sales['sunday'] = 2190

# Print the dictionary to verify
print(daily_sales)
````

### Delete a Key-Value Pair

Delete a value using the `del` keyword, the dictionary name, and the key.

````python
# Print the dictionary to verify
print(daily_sales)

# Delete a key-value pair
del daily_sales['sunday']

# Print the dictionary to verify
print(daily_sales)
````

## Accessing Dictionaries

We can access keys, and values, both at once, as well as loop through them.

````python
# Declaring the dictionary again to ensure it exists
sales_by_store = {
    "New York" : 72450,
    "Paris" : 65948,
    "Peckham" : 243763
}

# Retrieve an individual value by calling it's associated key
print(sales_by_store['Peckham'])

# Try expanding the 'sales_by_store' dictionary, and practice accessing values
````

Accessing individual values is useful for simple queries, but we may want to return all of the keys or values, for which we have the `keys()` and `values()` methods:

````python
print(sales_by_store.keys())

print(sales_by_store.values())
````

### Looping Through a Dictionary

The previous code simply lists out the keys and values, which can be useful when you need to identify what keys and values exist in the dictionary, but the methods are much more valuable when used within a loop.

````python
for key in sales_by_store.keys():
    print(f'the key is {key}')
    
for value in sales_by_store.values():
    print(f'the value is {value}')
````

In addition to looping through the **keys** and **values**, we may also want to loop through both, which can also be done with a **for loop**, but we need to use the `items()` method, and there is one additional consideration.

So far our for loops have required one temporary variable, when we loop through a list, a tuple, keys or values, we are working with one piece of data at a time. To loop through keys and values together, we require two temporary variables, one for the key, one for the value.

````python
for key, value in sales_by_store.items():
    print(f"Sales for {key} were {value}")

print("---\nAll stores reported")
````

### List Comprehension with Dictionaries

We can also use list comprehension to loop through dictionaries when we want to add values to a list.

````python
# Estimate quarterly sales

quarterly_sales = [value/4 for value in sales_by_store.values()]
print(quarterly_sales)
````

## If statements

So far our lines of code have been processed in order, from the first line to the last, however we quickly encounter limitations. For example, what if we only want to carry out operations against selected items in our lists or dictionaries, so far we've done the same thing to every piece of data.

**If statements** allow us to carryout checks or tests in our code, then perform different actions depending upon the result of the check or test. If the outcome is `True` then do one thing, if the outcome is `False` do something else.

The most commonly used tests will utilise the comparison operators that we reviewed in lesson 1, but more complex tests can be created. However, the output of your test should always be a boolean (True/False) value.

````python
average_temp = 30
todays_temp = 34

if todays_temp > average_temp:
    print("Today's temperature was above average")
````

In the above example our line of code our print statement runs when the test is `True`; We can use `if` and `else` statements to run a different lines of code when the outcome is `False`.

````python
average_temp = 30
todays_temp = 24

if todays_temp > average_temp:
    print("Today's temperature was above average")
else:
    print("Today's temperature was below average")
````

We can create more than two possible outputs for our if statements by using `elifs` (else-if)

````python
average_temp = 30
todays_temp = 31

if todays_temp > average_temp:
    print("Today's temperature was above average")
elif todays_temp == average_temp:
    print("Today's temperature was average")
else:
    print("Today's temperature was below average")
````

You can add as many `elif`'s as you want

### If Statements with For Loops

Above the if statement looks at one value, but we can embed it into a for loop to perform a comparison against several items in an iterable.

````python
daily_sales = [2345, 3254, 1123, 3546, 6345, 8562, 1996]
# Sales target is between 2000-4000

for value in daily_sales:
    if value < 2000:
        print(f"{value} is below target")
    elif value <= 4000:
        print(f"{value} is within target")
    elif value > 4000:
        print(f"{value} is above target")
````

>**Notice**: The else statement is not required, it is usually used as a fallback response when all previous checks are `False` 

### Logical Operators

Used in many places, but particularly in if statements, logical operators allow us to combine multiple conditional operators to create more complex checks.

|Operator|Description|Example|
|---|---|---|
|`and`|true if both conditions are true|`x > 5 and x < 10`|
|`or`|true if at least one condition is true|`name == 'Fred' or name == 'Ahmed'`|
|`not`|reverse the result i.e. if `x` exists return false, and vice-versa|`not x`|

A common tool to understand logical operators is to look at [Binary Truth Tables](https://www.realdigital.org/doc/e127ebfa82dbc904b5c0dac5d1adce8e)

Here's a simple example:

````python
age = 17
film_rating = 15

if (age < 18 and film_rating == 18): # Both conditions must be true
    print("You're too young, this film might disturb you!")
else:
    print("Come on in, the trauma is on you!")
````

Below we use logical operators in a multi-level **if-elif** statement, nested within a **for loop**.

````python
test_scores = [56, 64, 49, 71, 79, 45, 83, 60, 53, 88, 69, 91]

for result in test_scores:
    if result < 50:
        print(f"{result} is a fail")
    elif result >= 50 and result < 60:
        print(f"{result} is a grade D")
    elif result >= 60 and result < 70:
        print(f"{result} is a grade C")
    elif result >= 70 and result < 80:
        print(f"{result} is a grade B")
    elif result >= 80 and result < 90:
        print(f"{result} is a grade A")
    elif result >= 90:
        print(f"{result} is a grade A*")
````

### If Statements with List Comprehensions

If statements can be used to make our list comprehension statements more powerful by testing the values in the iterable.

The dictionary below contains the scores students achieved in a test, the passing score is 70. A list comprehension can be used to create a list of names of the students who passed.

````python
scores = {"Alice": 85, "Bob": 58, "Roisin": 68, "Charlie": 92, "Diana": 45, "Ahmed": 75}

# Extract names of students scoring 70 or higher
passed_students = [name for name, score in scores.items() if score >= 70]

print(passed_students)
# Expected output ["Alice", "Charlie", "Ahmed"]
````
