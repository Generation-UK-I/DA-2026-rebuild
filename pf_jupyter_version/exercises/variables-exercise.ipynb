## Python Assert

Each of the exercises above, and the upcoming ones use a Python function called `assert` to verify that you have completed the task successfully. If you look up the word 'assertion' in a dictionary, you'll find it is a statement which is believed to be true, and that's basically what it does in Python too, although in Python it's `True` not "true".

`assert` verifies that something you have defined is True, and if not then an `AssertionError` is raised. It is most commonly used for debugging and testing.

Let's look at one of the asserts that have been used above:

```py
assert myVar == 25, f"Expected: 25, but got: {myVar}."
print("Correct! Great job.")
```

Breakdown:

- `assert myVar == 25`: The assert statement is claiming that the myVar variable should be equal to the interger `25`
- `,`: After the assert statement is a comma which is optional, but allows us to provide an assertion message
- `f"Expected: 25, but got: {myVar}."`: This message is optional, but recommended. It is displayed if the assert statement fails, and should provide insight into why it failed.
- `print("Correct! Great job.")`: If the assert statement passes then this is the next line of code which runs. If the statement fails with an AssertionError this line is not reached.
