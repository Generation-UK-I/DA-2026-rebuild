## String Concatenation

Concatenation is a techy word for linking things together; If we add two numbers together we get a new number, but if we add two strings together (concatenate them) we get a new longer string, comprised of the originals.

```python
a = "hello "
b = "world"
print(a + b)

# the below code does the same thing using an 'assignment operator'
c = "goodbye "
c += "moon"
print(c)
```

Concatenation allows us to combine several short strings into one longer one.

```python
a = "quantity"
b = "time"

print("The line graph axis are " + a  + " and " + b)
```

>**Notice**: Spaces are included in the string, there is one after "are " and around the word " and ".

## String Interpolation (f-strings)

String concatenation involves making a new longer string by adding shorter ones together, but there is an alternative more powerful method that we can use to provide the same output, and more, called f-strings (formatted strings).

The below example will print the same message twice, once using concatenation, and once using an **f-string**.

```python
a = "quantity"
b = "time"

print("The line graph axis are " + a  + " and " + b)
print(f"The line graph axis are {a} and {b}")
```

F-strings are a more modern approach to working with strings, with a few advantages:

- They allow you to embed your variables without constantly opening and closing your strings
- They don't require you to specify or change the data type of variables

Here is the last concatenation example:

```python
a = "Frankie"
b = "Scout"
age_a = 6
age_b = 3

print(a + " is " + str(age_a) + " and " + b + " is " + str(age_b))
```

Here is the same code using an f-string:

```python
a = "Frankie"
b = "Scout"
age_a = 6
age_b = 3

print(f"{a} is {age_a} and {b} is {age_b}") # This one succeeds
```

With concatenation you need to explicitly convert the numeric values into strings using the `str()` method, but when using an f-string you did not, this is because f-strings carry out interpolation, not concatenation.

**Interpolation** means "*inserting something new*". 

Think of f-strings like filling in the blanks in your main string with any variable you want, whereas concatenation is joining separate individual strings together.

## String Methods

There is one more important concept to understand with strings, although it applies to many different object types in Python, called **methods**.

Methods are built in functionality available to the objects we create, there are various different methods available for strings, with additional methods available for lists, dictionaries, and many other Python objects.

Below are some examples of string methods:

```python
name = "rick james"

print(name.upper()) # Convert to uppercase
print(name.title()) # Capitalise every word
print(name.split(" ")) # Split at the specified separator and return a list
```
