## Stretch and Challenge

Write a function which calculates average daily sales; It should receive 5 values which are entered by the user during execution.

```python
daily_sales = []

for x in range(1,6):
    value = input('Enter the daily sales total:')
    daily_sales.append(value)

def calc_avg(sales):
    total = 0
    for x in sales:
        total += int(x)
    average = total/len(daily_sales)
    return average

daily_average = calc_avg(daily_sales)

print(daily_average)
```
