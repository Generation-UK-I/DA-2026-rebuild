## What is Python?

Python is a high-level programming language, which is accessible for beginners, but very powerful and flexible through the use of plugin modules, making it very popular for working with AI models, data science applications, and many other areas.

>Python is an interpreted language, which means when you run it the Python *interpreter* reads each line of code, in sequence, and translates the code into operations that the computer can process.

### What is Programming

Programming is the act of writing commands i.e. lines of code, which instruct the computer to carry out an operation. Computers operate on binary (1s and 0s), and of course humans don't, therefore we need abstraction layers to translate the instructions we type using Python commands, into binary operations the computer can run; we also rely upon translation layers in the other direction, to take binary output, and turn it into something humans can understand.

We can write programs to carry out a range of different operations, however they all share some concepts. They need to:

- Accept input, from the user, from events, from API calls, and more.
- Process the input, which may involve operating on the input, or it may be a trigger to process other elements.
- Access storage: fast RAM for the duration of the application's runtime; persistent storage if data needs to outlive the application.
- Produce some form of output. The output could be text based, graphics, audio, it may just be a confirmation, error message, or even just a status code.
- When working with **data**, we will usually be expecting either numerical output, or graphical output in the form of data visualisations.

## Operators

Operators allow us to quickly carry out actions with or against our variables and hardcoded values.

We can utilise a range of built in operators, examples of some of the most common **arithmetic** and **comparison** operators and how they may be used can be seen below.

Copy/paste the below code into your myApp.py file, run it and review the output, then try changing the values of `a` and `b` and re-running to verify the operators work as you expect.

>**Notice**: By assigning values to variables you only need to change them once and all of the operations against those variables can immediately utilise the new values.

````python
a = 15
b = 10

# Print data types
print(type(a))
print(type(b))

# Arithmetic Operators - returns values
print(a + b)
print(a - b)
print(b / a)
print(a * b)
print(a ** b)
print(b % a)

# Comparison Operators - returns boolean
print(a == b)
print(a != b)
print(a > b)
print(a >= b)
print(a < b)
print(a <= b)
````

>**Notice**: The hash symbol (`#`) is used to add comments to our code, these lines are messages to humans looking at our code, usually explaining its purpose, or for students often giving hints and tips. Python doesn't try to execute them.

The below operators are used to quickly update the value of a variable:

````python
my_var = 10

# Three ways to add 5 to the my_var variable:
# 1.
add_num = 5
new_val = my_var + add_num
print(new_val)

# 2.
my_var = my_var + 5
print(my_var)

# 3. Using an Assignment operator
my_var += 5
print(my_var)

# More assignment operators:
my_var += 5 # add 5 to my_var
my_var -= 5 # subtract 5 from my_var
my_var *= 5 # multiple my_var by 5
my_var /= 5 # divide my_var by 5
my_var %= 5 # return the modulus of my_var / 5
my_var **= 5 # raise my_var to the power of 5
````

Try testing the above operators using your own values - you may need to add extra `print()` statements.

## Working with Jupyter Notebooks

Unlike a traditional Python script (.py file), a notebook (.ipynb) is divided into cells.

### Notebook Cells

Each cell can contain either documentation or executable code, making notebooks ideal for teaching, experimentation, and data analysis.

A notebook typically contains two main cell types, markdown and code.

### Markdown Cells

Mardown is a lightweight markup language that allows you to format plain text using simple keyboard symbols. Markdown cells are used for notes, explanations, instructions, and documentation.

Create your first markdown cell by clicking the markdown button which appears at the top of your first notebook.

![md_cell](./imgs/md_cell.png)

Create a markdown cell and type the following exactly as it appears:

```markdown
# This is a Markdown Heading

## This is a Sub-Heading

Markdown can render in many different interpreters including VSC, in Jupyter Notebooks it will render when you execute the markdown cell.
```

Once entered press `CTRL + ENTER` and your Markdown will be rendered and will now look like the text below.

---

# This is a Markdown Heading

## This is a Sub-Heading

Markdown can render in many different interpreters, in Jupyter Notebooks it will render when you execute the markdown cell.

---

There will be a lot of markdown blocks throughout these Python modules, feel free to double click on them in order to learn the syntax used to create them.

### Code Cells

Code cells can be used to run our code directly in the document, with any output displaying directly below the cell. In our case we're running Python, but many languages are supported.

If you hover your mouse near the bottom of your existing markdown cell you will see the option to create a code cell, when you click it a new cell is created, but it has a 'play' button to the left, which you can use to run the code within - if your cursor is in the cell you can also execute the code with `CTRL + ENTER`.

![code_cell](./imgs/code_cell.png)

There is one final requirement to get your Jupyter environment up and running, create a code cell, enter print('Hello World'), then execute it. When you do you'll be prompted to install the `ipykernel package`, click `install`.

![install_ipykernel](./imgs/ipykernel_install.png)

Ipykernel is the specific Python environment which will run inside your notebook.

---

Both code and markdown cells can be executed independently using:

- `Shift + Enter` → Run cell and move to next
- `Ctrl + Enter` → Run cell and stay on current cell
- For code cells only, you can also click the 'play' button to the left of the cell
