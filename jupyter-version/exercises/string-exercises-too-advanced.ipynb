## Install Pytest and Ipytest

In order to verify that your exercise answers are correct we need to run some tests. To ensure that the tests run correctly we need to intall a couple of additional Python libraries. 

If you do not have a Terminal window open, at the top of VSC open the 'Terminal' menu (it may be hidden under 3 dots `...`) and select 'New Terminal'.

In the terminal type: `pip install pytest ipytest`

- **pip**: Python's package manager
- **pytest**: A common python testing framework
- **ipytest**: Acts as a bridge between pytest and the embedded code environment of Jupyter Notebooks

If successful you should see a message like:

`Successfully installed iniconfig-2.3.1 ipytest-0.14.2 pluggy-1.6.0 pytest-9.1.1`
