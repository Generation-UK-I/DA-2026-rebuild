## Python Libraries

A default installation of Python includes a wide range of functionality, however many of it's features are modularised and not available immediately. This keeps your installation lean and fast, because it only loads what you need. To access the extra functionality we can import it using an `import module_name` statement.

Many different modules are available out-of-the-box these are referred to as the Python Standard Library. However, there are hundreds of additional modules which may be installed manually.

We're going to direct you to complete some practice exercises shortly, for which you will need to install two additional libraries, called `pytest` and `ipytest`.

## Unit Testing

At its core, unit testing is the practice of writing automated code to verify that individual pieces of your code behave exactly as expected under various conditions.

There are a range of ways to carry out testing, and as with most things, ideally we want to automate it. One of the most common testing frameworks is called `pytest`.

## Functions

The last feature we're giving you a sneak peak into is Functions, which is one of the standard approaches to writing code. Functions allow you to write individual code blocks which perform a specific action, and separate your code blocks so that they can be called when required, instead of running one line at a time sequentially from top to bottom.

### Example

A function is created with the def keyword, followed by a descriptive function name, for example:

```py
def hello_world():
```

When you run the function it executes code that is indented within the function:

```py
def hello_world():
    print('hello world')
```

Notice the lines of code to be executed are indentated from the function definition by 1x `TAB`.

In order to exectute the function we need to call it, which we do by stating the function name:

```py
def hello_world():
    print('hello world')

hello_world()
```

You might be asking "why bother?" that is more lines than just `print('hello world')`, but the function can contain many lines of code, and it can be called over and over again:

```py
def hello_world():
    print('hello world')
    print('this is a print statement')
    print('inside a function)

hello_world()
hello_world()
hello_world()
```
